```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score


data = pd.read_csv("data.csv")


X = data.select_dtypes(include=["int64", "float64"]).copy()


X = X.drop(["SalePrice", "Id"], axis=1)

X = X.fillna(X.mean())


y = data["SalePrice"].values


print("X shape:", X.shape)
print("y shape:", y.shape)
print("چند NaN باقی موند؟\n", X.isnull().sum())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X.values, y, test_size=0.2, random_state=42
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python

X_train_scaled = np.c_[np.ones(X_train_scaled.shape[0]), X_train_scaled]
X_test_scaled = np.c_[np.ones(X_test_scaled.shape[0]), X_test_scaled]

print("بعد از اضافه کردن bias:", X_train_scaled.shape)
```

```python
def gradient_descent(X, y, lr=0.01, iterations=3000):
    m, n = X.shape
    theta = np.zeros(n)

    for _ in range(iterations):
        y_pred = X.dot(theta)
        error = y_pred - y
        gradient = (1/m) * X.T.dot(error)
        theta = theta - lr * gradient

    return theta
```

```python
theta = gradient_descent(X_train_scaled, y_train)
print("theta آموزش دیده:", theta)
```

```python
y_pred = X_test_scaled.dot(theta)

print("۵ پیش‌بینی اول:", y_pred[:5])
```

```python
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("MSE:", mse)
print("RMSE:", rmse)
print("R2:", r2)
```
